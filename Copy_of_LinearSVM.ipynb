```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
dataset = pd.read_csv('/content/Social_Network_Ads_400_rows.csv')
X = dataset.iloc[:, :-1].values
y = dataset.iloc[:, -1].values
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(X,y,test_size = 0.25,random_state = 0)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

```python
print(x_train)
```

```python
from sklearn.svm import SVC
classifier = SVC(kernel = 'rbf', random_state = 0)
classifier.fit(x_train, y_train)
```

```python
y_pred = classifier.predict(sc.fit_transform([[30,149734]]))
```

```python
print(y_pred)
```

```python
yp = classifier.predict(x_test)
```

```python
print(yp)
```

```python
ypred = classifier.predict(x_test)
print(np.concatenate((ypred.reshape(len(ypred),1),y_test.reshape(len(y_test),1)),1))
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score
confusion_matrix(y_test,yp)
```

```python
accuracy_score(y_test,yp)
```
